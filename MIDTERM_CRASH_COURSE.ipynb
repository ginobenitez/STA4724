# 📈 PART 2: LINEAR REGRESSION CRASH COURSE

## The Most Important Formula in This Course:
$$y = \beta_0 + \beta_1 x + \varepsilon$$

**Translation**: Outcome = Intercept + Slope × Input + Error

**Key Point**: This predicts CONTINUOUS values (like height, price, temperature)


````python
# Quick example with fake data
np.random.seed(42)
X = np.random.randn(100, 1) * 2  # Input variable
y = 3 + 2 * X.ravel() + np.random.randn(100) * 0.5  # y = 3 + 2x + noise

# Fit the model (THIS IS THE PATTERN!)
model = LinearRegression()
model.fit(X, y)

print(f"True intercept: 3, Fitted: {model.intercept_:.2f}")
print(f"True slope: 2, Fitted: {model.coef_[0]:.2f}")

# Plot it
plt.figure(figsize=(8, 6))
plt.scatter(X, y, alpha=0.6, label='Data points')
plt.plot(X, model.predict(X), 'r-', linewidth=2, label='Fitted line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression: y = 3 + 2x + noise')
plt.legend()
plt.show()

print("✅ You just did linear regression!")
````

### 🎯 KEY REGRESSION CONCEPTS (EXAM FAVORITES)

1. **Ordinary Least Squares (OLS)**: Find the line that minimizes squared errors
2. **R²**: How much variance does your model explain? (Higher = better, 0-1 scale)
3. **Overfitting**: Model too complex, fits noise (bad)
4. **Underfitting**: Model too simple, misses patterns (also bad)
5. **Bias-Variance Trade-off**: Balance between these two problems


````python
# Calculate R² score (WILL BE ON EXAM)
from sklearn.metrics import r2_score
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)
print(f"R² Score: {r2:.3f}")
print(f"This means our model explains {r2*100:.1f}% of the variance")

# The closer R² is to 1.0, the better your model!
````

# 🎯 PART 3: CLASSIFICATION (kNN) CRASH COURSE

**Big Idea**: Classify new points based on their k nearest neighbors

**Key Difference from Regression**: This predicts CATEGORIES (like species, pass/fail, spam/not spam)

**From your HW3**: This is exactly what you need to do!


````python
# Let's use the classic Iris dataset (simple and clean)
from sklearn.datasets import load_iris

iris = load_iris()
X, y = iris.data, iris.target

print(f"Dataset shape: {X.shape}")
print(f"Classes: {iris.target_names}")
print("Features:", iris.feature_names)
print("\nFirst 5 rows of data:")
print(X[:5])
print("First 5 labels:")
print(y[:5])
````

````python
# Split data (ALWAYS do this!)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Try different k values (this is EXACTLY what HW3 asks!)
k_values = [1, 3, 5, 7, 9]
scores = []

print("Testing different k values:")
for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    score = knn.score(X_test, y_test)
    scores.append(score)
    print(f"k={k}: Accuracy = {score:.3f}")

# Find best k
best_k = k_values[np.argmax(scores)]
print(f"\n🏆 Best k = {best_k} with accuracy = {max(scores):.3f}")
print("✅ This is cross-validation in action!")
````

````python
# Confusion Matrix (DEFINITELY on the exam)
best_knn = KNeighborsClassifier(n_neighbors=best_k)
best_knn.fit(X_train, y_train)
y_pred = best_knn.predict(X_test)

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
print("\nHow to read this:")
print("- Rows = True labels")
print("- Columns = Predicted labels") 
print("- Diagonal = Correct predictions")

# Manual accuracy calculation (HW3 requirement!)
correct_predictions = np.sum(np.diag(cm))
total_predictions = np.sum(cm)
manual_accuracy = correct_predictions / total_predictions

print(f"\n📊 Manual accuracy calculation:")
print(f"Correct predictions: {correct_predictions}")
print(f"Total predictions: {total_predictions}")
print(f"Accuracy = {correct_predictions}/{total_predictions} = {manual_accuracy:.3f}")

# Verify with sklearn
sklearn_accuracy = accuracy_score(y_test, y_pred)
print(f"Sklearn accuracy (should match): {sklearn_accuracy:.3f}")
print("✅ Perfect match!" if abs(manual_accuracy - sklearn_accuracy) < 0.001 else "❌ Something's wrong")
````

# 🎓 PART 4: MODEL EVALUATION ESSENTIALS

**Cross-Validation**: Test your model's performance more reliably than a single train/test split

**Why it matters**: Single splits can be unlucky - CV gives you a better estimate


````python
# Cross-validation (HW3 asks for this!)
from sklearn.model_selection import cross_val_score

# Test k=3 with 5-fold cross-validation
knn = KNeighborsClassifier(n_neighbors=3)
cv_scores = cross_val_score(knn, X, y, cv=5)

print(f"Cross-validation scores: {cv_scores}")
print(f"Mean CV score: {cv_scores.mean():.3f} ± {cv_scores.std():.3f}")
print(f"This is more reliable than a single train/test split!")

# Compare different k values with CV
print("\n🔍 Finding best k with cross-validation:")
for k in [1, 3, 5, 7, 9]:
    knn = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(knn, X, y, cv=5)
    print(f"k={k}: CV accuracy = {cv_scores.mean():.3f} ± {cv_scores.std():.3f}")
````

### 3. 🔧 Two Ways to Do Linear Regression (HW2 Requirement)


````python
# Prepare data
X = df[['age']].values  # Input features (as 2D array)
y = df['cost'].values   # Target variable

# METHOD 1: Using sklearn (the easy way)
model_sklearn = LinearRegression()
model_sklearn.fit(X, y)

print("METHOD 1 - Sklearn:")
print(f"Intercept (β₀): {model_sklearn.intercept_:.2f}")
print(f"Slope (β₁): {model_sklearn.coef_[0]:.2f}")

# METHOD 2: Manual calculation using closed-form solution
# Formula: β = (X^T X)^(-1) X^T y
# For simple regression: β₁ = Σ(x-x̄)(y-ȳ) / Σ(x-x̄)²

# Add intercept column (column of 1s)
X_with_intercept = np.column_stack([np.ones(len(X)), X])

# Closed-form solution: β = (X^T X)^(-1) X^T y
XtX = X_with_intercept.T @ X_with_intercept
XtX_inv = np.linalg.inv(XtX)
Xty = X_with_intercept.T @ y
beta_manual = XtX_inv @ Xty

print(f"\nMETHOD 2 - Manual calculation:")
print(f"Intercept (β₀): {beta_manual[0]:.2f}")
print(f"Slope (β₁): {beta_manual[1]:.2f}")

print(f"\n✅ Both methods should give same result!")
print(f"Sklearn vs Manual difference: {abs(model_sklearn.intercept_ - beta_manual[0]):.6f}")
````

### 4. 🎨 Plot Both Fitted Lines (HW2 Always Asks)


````python
# Plot data and both fitted lines
plt.figure(figsize=(10, 6))

# Original data
plt.scatter(df['age'], df['cost'], alpha=0.7, s=50, label='Data', color='blue')

# Sklearn predictions
y_pred_sklearn = model_sklearn.predict(X)
plt.plot(df['age'], y_pred_sklearn, 'r-', linewidth=2, label='Sklearn fit')

# Manual predictions  
y_pred_manual = X_with_intercept @ beta_manual
plt.plot(df['age'], y_pred_manual, 'g--', linewidth=2, label='Manual fit')

plt.xlabel('Tractor Age (years)')
plt.ylabel('Maintenance Cost ($)')
plt.title('Linear Regression: Two Methods')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("✅ Both lines should be identical (or very close)!")

# Make a prediction (HW2 always asks this)
new_age = 3
pred_cost = model_sklearn.predict([[new_age]])[0]
print(f"\n🔮 Prediction: A {new_age}-year-old tractor will cost ${pred_cost:.2f} for maintenance")
````

### 5. 🎯 Lasso Regression (Advanced Topic from HW2)


````python
# Lasso Regression (regularized regression)
from sklearn.linear_model import Lasso
from sklearn.model_selection import cross_val_score

# Generate synthetic data like HW2 Question 2
np.random.seed(0)
m, n = 150, 75  # 150 samples, 75 features

# Random feature matrix
X_synthetic = np.random.rand(m, n)

# True coefficients (sparse - only first 10 are non-zero)
beta_true = np.zeros(n)
beta_true[:10] = np.random.uniform(-10, 10, 10)  # Random values between -10 and 10

# Generate target with noise
noise = np.random.randn(m) * 0.1
y_synthetic = X_synthetic @ beta_true + noise

print(f"Synthetic data shape: X={X_synthetic.shape}, y={y_synthetic.shape}")
print(f"True non-zero coefficients: {np.sum(beta_true != 0)}")

# Fit Lasso with specific lambda (alpha in sklearn)
lasso = Lasso(alpha=0.00001)  # lambda = 0.00001 from HW2
lasso.fit(X_synthetic, y_synthetic)

print(f"\nLasso found {np.sum(lasso.coef_ != 0)} non-zero coefficients")
print(f"Distance from true β: {np.linalg.norm(lasso.coef_ - beta_true):.4f}")

# Cross-validation to find best alpha (HW2 asks for this)
alphas = np.logspace(-6, 0, 50)  # Test different regularization strengths
cv_scores = []

for alpha in alphas:
    lasso_cv = Lasso(alpha=alpha)
    scores = cross_val_score(lasso_cv, X_synthetic, y_synthetic, cv=10)
    cv_scores.append(scores.mean())

best_alpha = alphas[np.argmax(cv_scores)]
print(f"Best alpha from CV: {best_alpha:.6f}")

# Fit with best alpha
lasso_best = Lasso(alpha=best_alpha)
lasso_best.fit(X_synthetic, y_synthetic)
print(f"Best model distance from true β: {np.linalg.norm(lasso_best.coef_ - beta_true):.4f}")

print("✅ Lasso regression complete!")
````

### 6. 📊 Multivariate Regression (Multiple Features)


````python
# Multivariate regression (like HW2 Question 3 with beef data)
# Create fake beef data with SIZE and log(VAL) predicting YES%

beef_data = {
    'SIZE': [100, 150, 200, 250, 300, 350, 400, 450, 500, 550],
    'VAL': [50, 75, 100, 125, 150, 175, 200, 225, 250, 275],
    'YES': [45, 52, 58, 63, 67, 70, 72, 74, 75, 76]
}
beef_df = pd.DataFrame(beef_data)

# Apply log transformation (HW2 requirement)
beef_df['log_VAL'] = np.log(beef_df['VAL'])

print("Beef voting data:")
print(beef_df.head())

# Multivariate regression: YES ~ SIZE + log(VAL)
X_multi = beef_df[['SIZE', 'log_VAL']].values
y_multi = beef_df['YES'].values

multi_model = LinearRegression()
multi_model.fit(X_multi, y_multi)

print(f"\nMultivariate Regression Results:")
print(f"Intercept: {multi_model.intercept_:.3f}")
print(f"SIZE coefficient: {multi_model.coef_[0]:.6f}")
print(f"log(VAL) coefficient: {multi_model.coef_[1]:.3f}")

# Model evaluation
r2_multi = multi_model.score(X_multi, y_multi)
print(f"R² score: {r2_multi:.3f}")

# Make predictions
y_pred_multi = multi_model.predict(X_multi)

print(f"\n✅ Multivariate regression complete!")
print(f"💡 Model equation: YES = {multi_model.intercept_:.2f} + {multi_model.coef_[0]:.4f}×SIZE + {multi_model.coef_[1]:.2f}×log(VAL)")
````
