### Question 2: kNN Classification (HW3 Pattern)
**Complete this code - it's exactly like HW3:**


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np

# Load data
iris = load_iris()
X, y = iris.data, iris.target

# TODO: Split the data (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# TODO: Create and fit kNN classifier with k=3
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# TODO: Make predictions and calculate accuracy
y_pred = knn.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# TODO: Create confusion matrix and calculate accuracy manually
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Manual accuracy calculation (HW3 asks for this!)
correct_predictions = np.sum(np.diag(cm))
total_predictions = np.sum(cm)
manual_accuracy = correct_predictions / total_predictions

print(f"Manual accuracy: {correct_predictions}/{total_predictions} = {manual_accuracy:.3f}")
print(f"Should match sklearn: {accuracy:.3f}")
print("✅ Perfect!" if abs(manual_accuracy - accuracy) < 0.001 else "❌ Check your calculation")
```

### Question 3: Linear Regression (HW2 Pattern)
**Both sklearn AND manual methods:**


```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Generate simple data (like HW2 tractor example)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

print("Data:")
print("X:", X.ravel())
print("y:", y)

# METHOD 1: sklearn
model = LinearRegression()
model.fit(X, y)
print(f"\nSklearn method:")
print(f"Intercept: {model.intercept_:.3f}")
print(f"Slope: {model.coef_[0]:.3f}")

# METHOD 2: Manual calculation (HW2 requirement!)
X_with_intercept = np.column_stack([np.ones(len(X)), X])
XtX = X_with_intercept.T @ X_with_intercept
XtX_inv = np.linalg.inv(XtX)
Xty = X_with_intercept.T @ y
beta_manual = XtX_inv @ Xty

print(f"\nManual method:")
print(f"Intercept: {beta_manual[0]:.3f}")
print(f"Slope: {beta_manual[1]:.3f}")

# Should be identical!
print(f"\n✅ Methods match!" if abs(model.intercept_ - beta_manual[0]) < 0.001 else "❌ Check calculation")

# Make a prediction (always asked!)
new_x = 6
prediction = model.predict([[new_x]])[0]
print(f"\nPrediction for X={new_x}: {prediction:.1f}")

# What should this be? y = 0 + 2*x, so y = 2*6 = 12
```

```python
# 🔍 DETAILED EXPLANATION: Why c changes when a changes

# Let's trace through this step by step with memory addresses
a = [1, 2, 3]
b = [1, 2, 3] 
c = a

print("🧠 MEMORY ANALYSIS:")
print(f"a = {a}, memory location: {id(a)}")
print(f"b = {b}, memory location: {id(b)}")
print(f"c = {c}, memory location: {id(c)}")

print(f"\n🔑 KEY INSIGHT:")
print(f"a and c point to the SAME object: {id(a) == id(c)}")
print(f"a and b point to DIFFERENT objects: {id(a) == id(b)}")

print(f"\n📝 What 'c = a' really means:")
print(f"c = a does NOT copy the list")
print(f"c = a makes c point to the SAME list object as a")
print(f"Think of it as: c and a are two names for the same box")

# Now let's see what happens when we modify a
print(f"\n🔄 BEFORE appending:")
print(f"a = {a}")
print(f"c = {c}")

a.append(4)

print(f"\n🔄 AFTER a.append(4):")
print(f"a = {a}")
print(f"c = {c}")
print(f"b = {b}")  # b is unchanged because it's a different object

print(f"\n💡 WHY c changed:")
print(f"Because c and a point to the SAME list object,")
print(f"when you modify that object through 'a',")
print(f"you see the change through 'c' too!")
print(f"It's like having two keys to the same house.")
```

```python
# 🏠 HOUSE ANALOGY (This will help you remember!)

print("🏠 THINK OF IT LIKE HOUSES AND ADDRESSES:")
print()
print("a = [1, 2, 3]  →  Build a house at address 12345, put [1,2,3] inside")
print("b = [1, 2, 3]  →  Build a DIFFERENT house at address 67890, put [1,2,3] inside") 
print("c = a          →  Give c the address 12345 (same house as a!)")
print()
print("Now a and c both have keys to the SAME house (address 12345)")
print("b has keys to a DIFFERENT house (address 67890)")
print()
print("When you do a.append(4):")
print("- You go to house 12345 and add 4 to the contents")
print("- Now house 12345 contains [1,2,3,4]")
print("- Since both a and c point to house 12345, they both see [1,2,3,4]")
print("- b points to house 67890, which is unchanged: [1,2,3]")

print("\n" + "="*60)
print("🎯 EXAM TIP: This is why 'is' vs '==' matters!")
print("a == b  →  True (same contents)")
print("a is b  →  False (different houses/objects)")
print("a is c  →  True (same house/object)")
print("="*60)
```

```python
# 🛠️ HOW TO AVOID THIS (Create Independent Copies)

print("🛠️ IF YOU WANT INDEPENDENT LISTS:")
print()

# Method 1: Use .copy()
a1 = [1, 2, 3]
c1 = a1.copy()  # Creates a new list with same contents

print("Method 1: Using .copy()")
print(f"a1 = {a1}, id = {id(a1)}")
print(f"c1 = {c1}, id = {id(c1)}")
print(f"Same contents? {a1 == c1}")
print(f"Same object? {a1 is c1}")

a1.append(4)
print(f"After a1.append(4): a1={a1}, c1={c1}")
print("✅ c1 is unchanged because it's a separate copy!")

print("\n" + "-"*50)

# Method 2: Use slicing [:]
a2 = [1, 2, 3]
c2 = a2[:]  # Creates a new list with same contents

print("Method 2: Using slicing [:]")
print(f"a2 = {a2}, id = {id(a2)}")
print(f"c2 = {c2}, id = {id(c2)}")

a2.append(4)
print(f"After a2.append(4): a2={a2}, c2={c2}")
print("✅ c2 is unchanged because it's a separate copy!")

print("\n" + "-"*50)

# Method 3: Use list() constructor
a3 = [1, 2, 3]
c3 = list(a3)  # Creates a new list with same contents

print("Method 3: Using list() constructor")
print(f"a3 = {a3}, id = {id(a3)}")
print(f"c3 = {c3}, id = {id(c3)}")

a3.append(4)
print(f"After a3.append(4): a3={a3}, c3={c3}")
print("✅ c3 is unchanged because it's a separate copy!")

print("\n🎯 REMEMBER FOR EXAM:")
print("c = a        →  Same object (changes together)")
print("c = a.copy() →  Different objects (independent)")
print("c = a[:]     →  Different objects (independent)")
print("c = list(a)  →  Different objects (independent)")
```

# 👶 LINEAR REGRESSION FOR BEGINNERS
## From Zero to Hero in Simple Steps!

**Goal**: Understand linear regression so well that you can explain it to a 5-year-old!


```python
## 🍎 STEP 1: What is Linear Regression? (5-Year-Old Explanation)

print("🍎 IMAGINE YOU'RE SELLING APPLES:")
print()
print("You notice a pattern:")
print("- When it's 70°F outside → You sell 10 apples")
print("- When it's 80°F outside → You sell 20 apples") 
print("- When it's 90°F outside → You sell 30 apples")
print()
print("🤔 QUESTION: If it's 85°F tomorrow, how many apples will you sell?")
print()
print("🧠 YOUR BRAIN THINKS:")
print("'Hmm, there's a pattern here...'")
print("'For every 10 degrees warmer, I sell 10 more apples!'")
print("'So 85°F should be... 25 apples!'")
print()
print("🎯 THAT'S LINEAR REGRESSION!")
print("You found the LINE that best fits your data!")
print("Temperature goes UP → Apple sales go UP")
print("It's a STRAIGHT LINE relationship!")

import numpy as np
import matplotlib.pyplot as plt

# Let's visualize this apple example
temperature = np.array([70, 80, 90])
apples_sold = np.array([10, 20, 30])

plt.figure(figsize=(8, 6))
plt.scatter(temperature, apples_sold, color='red', s=100, label='Your data')
plt.plot([70, 90], [10, 30], 'b-', linewidth=2, label='The pattern (line)')
plt.xlabel('Temperature (°F)')
plt.ylabel('Apples Sold')
plt.title('🍎 Apple Sales vs Temperature')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("👆 See that blue line? That's your LINEAR REGRESSION!")
```

```python
## 📐 STEP 2: The Magic Formula (Remember This Forever!)

print("📐 EVERY STRAIGHT LINE HAS THE SAME FORMULA:")
print()
print("🎯 y = mx + b")
print()
print("Where:")
print("📊 y = what you want to predict (apples sold)")
print("📈 x = what you know (temperature)")  
print("📐 m = slope (how steep the line is)")
print("📍 b = y-intercept (where line crosses y-axis)")
print()
print("🍎 FOR OUR APPLE EXAMPLE:")
print("Apples = m × Temperature + b")
print()
print("Let's figure out m and b...")

# Calculate slope (m) manually
x1, y1 = 70, 10  # First point
x2, y2 = 90, 30  # Second point

slope_m = (y2 - y1) / (x2 - x1)
print(f"\n📈 SLOPE (m) = (y2 - y1) / (x2 - x1)")
print(f"m = ({y2} - {y1}) / ({x2} - {x1})")
print(f"m = {y2 - y1} / {x2 - x1}")
print(f"m = {slope_m}")
print(f"👉 This means: For every 1°F increase, we sell {slope_m} more apples!")

# Calculate y-intercept (b) manually
# Using point (70, 10) and y = mx + b
# 10 = 1 × 70 + b
# b = 10 - 70 = -60
intercept_b = y1 - slope_m * x1
print(f"\n📍 Y-INTERCEPT (b) using point ({x1}, {y1}):")
print(f"{y1} = {slope_m} × {x1} + b")
print(f"b = {y1} - {slope_m} × {x1}")
print(f"b = {y1} - {slope_m * x1}")
print(f"b = {intercept_b}")

print(f"\n🎯 OUR FINAL FORMULA:")
print(f"Apples = {slope_m} × Temperature + ({intercept_b})")
print(f"or")
print(f"y = {slope_m}x + ({intercept_b})")

# Test our formula
test_temp = 85
predicted_apples = slope_m * test_temp + intercept_b
print(f"\n🧪 TEST: At {test_temp}°F:")
print(f"Apples = {slope_m} × {test_temp} + ({intercept_b})")
print(f"Apples = {slope_m * test_temp} + ({intercept_b})")
print(f"Apples = {predicted_apples}")
print(f"✅ We predict {predicted_apples} apples!")
```

```python
## 🤖 STEP 3: Let Python Do the Math (sklearn Method)

print("🤖 NOW LET'S USE PYTHON TO DO THE SAME THING:")
print("(This is what you'll use on your exam!)")

from sklearn.linear_model import LinearRegression

# Our data (same apple example)
X = np.array([[70], [80], [90]])  # Temperature (must be 2D for sklearn)
y = np.array([10, 20, 30])        # Apples sold

print(f"\n📊 Our data:")
print(f"X (temperature): {X.ravel()}")  # .ravel() makes it 1D for printing
print(f"y (apples): {y}")

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

print(f"\n🎯 SKLEARN RESULTS:")
print(f"Slope (coefficient): {model.coef_[0]}")
print(f"Y-intercept: {model.intercept_}")

print(f"\n🔍 COMPARE TO OUR MANUAL CALCULATION:")
print(f"Manual slope: {slope_m}")
print(f"Manual intercept: {intercept_b}")
print(f"✅ They match! (or very close)")

# Make predictions
test_temps = np.array([[75], [85], [95]])
predictions = model.predict(test_temps)

print(f"\n🧪 PREDICTIONS:")
for temp, pred in zip(test_temps.ravel(), predictions):
    print(f"At {temp}°F → {pred:.1f} apples")

# Calculate R² score (how good is our model?)
r2_score = model.score(X, y)
print(f"\n📊 R² SCORE: {r2_score:.3f}")
print(f"This means our model explains {r2_score*100:.1f}% of the variation!")
print(f"(1.0 = perfect, 0.0 = terrible)")
```

```python
## 🧮 STEP 4: The Matrix Way (HW2 Requirement!)

print("🧮 YOUR PROFESSOR WANTS YOU TO KNOW THE MATRIX METHOD:")
print("This is the 'manual' calculation that HW2 always asks for!")
print()
print("🎯 THE MAGIC FORMULA: β = (X^T X)^(-1) X^T y")
print("Don't worry about what it means - just learn the pattern!")

# Prepare data for matrix calculation
X_matrix = np.array([[70], [80], [90]])  # Our temperatures
y_vector = np.array([10, 20, 30])        # Our apple sales

print(f"\n📊 STEP 1: Add intercept column (column of 1s)")
# Add column of 1s for intercept
X_with_intercept = np.column_stack([np.ones(len(X_matrix)), X_matrix])
print(f"X with intercept:")
print(X_with_intercept)
print(f"Shape: {X_with_intercept.shape}")

print(f"\n🔢 STEP 2: Calculate X^T X (X transpose times X)")
XtX = X_with_intercept.T @ X_with_intercept  # @ means matrix multiplication
print(f"X^T X =")
print(XtX)

print(f"\n🔄 STEP 3: Calculate (X^T X)^(-1) (inverse)")
XtX_inverse = np.linalg.inv(XtX)
print(f"(X^T X)^(-1) =")
print(XtX_inverse)

print(f"\n🎯 STEP 4: Calculate X^T y")
Xty = X_with_intercept.T @ y_vector
print(f"X^T y =")
print(Xty)

print(f"\n✨ STEP 5: Final calculation β = (X^T X)^(-1) X^T y")
beta = XtX_inverse @ Xty
print(f"β =")
print(beta)
print(f"\nIntercept (β₀): {beta[0]}")
print(f"Slope (β₁): {beta[1]}")

print(f"\n🔍 COMPARE ALL THREE METHODS:")
print(f"Manual calculation:  slope={slope_m:.3f}, intercept={intercept_b:.3f}")
print(f"Sklearn:            slope={model.coef_[0]:.3f}, intercept={model.intercept_:.3f}")
print(f"Matrix method:      slope={beta[1]:.3f}, intercept={beta[0]:.3f}")
print(f"✅ All three methods give the SAME answer!")
```

```python
## 🏋️ PRACTICE TIME! (Do These Until You Can Do Them With Eyes Closed)

print("🏋️ PRACTICE PROBLEM 1: Ice Cream Sales")
print("You run an ice cream truck and notice:")
print("- At 60°F: You sell 5 ice creams")
print("- At 80°F: You sell 25 ice creams")
print("- At 100°F: You sell 45 ice creams")
print()
print("🎯 YOUR MISSION:")
print("1. Calculate slope and intercept manually")
print("2. Use sklearn to verify")
print("3. Use matrix method to double-check")
print("4. Predict sales at 90°F")
print()
print("👇 TRY IT YOURSELF FIRST, THEN RUN THE SOLUTION BELOW!")

# SOLUTION (try yourself first!)
print("\n" + "="*50)
print("🔍 SOLUTION:")

# Data
temps = np.array([60, 80, 100])
ice_creams = np.array([5, 25, 45])

# Method 1: Manual calculation
x1, y1 = 60, 5
x2, y2 = 100, 45
manual_slope = (y2 - y1) / (x2 - x1)
manual_intercept = y1 - manual_slope * x1

print(f"\n📐 Manual calculation:")
print(f"Slope = ({y2} - {y1}) / ({x2} - {x1}) = {manual_slope}")
print(f"Intercept = {y1} - {manual_slope} × {x1} = {manual_intercept}")

# Method 2: sklearn
X_ice = temps.reshape(-1, 1)  # Make it 2D
y_ice = ice_creams

model_ice = LinearRegression()
model_ice.fit(X_ice, y_ice)

print(f"\n🤖 Sklearn:")
print(f"Slope: {model_ice.coef_[0]}")
print(f"Intercept: {model_ice.intercept_}")

# Method 3: Matrix method
X_with_ones = np.column_stack([np.ones(len(X_ice)), X_ice])
beta_ice = np.linalg.inv(X_with_ones.T @ X_with_ones) @ X_with_ones.T @ y_ice

print(f"\n🧮 Matrix method:")
print(f"Slope: {beta_ice[1]}")
print(f"Intercept: {beta_ice[0]}")

# Prediction
temp_90 = 90
prediction_90 = model_ice.predict([[temp_90]])[0]
print(f"\n🧪 Prediction at {temp_90}°F: {prediction_90:.1f} ice creams")

print(f"\n✅ All methods should give the same answer!")
```
