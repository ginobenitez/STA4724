# 🎯 LECTURE 7: BIAS-VARIANCE TRADE-OFF

## The Most Important Concept in Machine Learning

print("🎯 BIAS-VARIANCE TRADE-OFF")
print("The secret to building great models!")
print()
print("🤔 THE PROBLEM:")
print("   Your model can fail in two ways:")
print("   1. Too simple → UNDERFITTING (high bias)")
print("   2. Too complex → OVERFITTING (high variance)")
print()
print("💡 THE SOLUTION:")
print("   Find the sweet spot in the middle!")
print()
print("🧠 SIMPLE ANALOGY:")
print("   🎯 Bias = Systematic error (like a broken compass)")
print("   📊 Variance = Random error (like shaky hands)")
print("   🎪 Goal = Low bias AND low variance")


```python
## 📊 Demonstrating Bias-Variance with Polynomial Regression

# Create a true underlying function (sine wave)
np.random.seed(42)
x_true = np.linspace(0, 1, 100)
y_true = np.sin(2 * np.pi * x_true)

# Generate noisy training data
n_samples = 20
x_train = np.random.uniform(0, 1, n_samples)
y_train = np.sin(2 * np.pi * x_train) + np.random.normal(0, 0.3, n_samples)

plt.figure(figsize=(15, 5))

# Model 1: Too simple (degree 1) - HIGH BIAS
plt.subplot(1, 3, 1)
plt.scatter(x_train, y_train, alpha=0.7, color='blue', label='Training data')
plt.plot(x_true, y_true, 'g-', linewidth=2, label='True function')

# Fit linear model
poly_1 = PolynomialFeatures(degree=1)
X_train_1 = poly_1.fit_transform(x_train.reshape(-1, 1))
X_true_1 = poly_1.transform(x_true.reshape(-1, 1))
model_1 = LinearRegression()
model_1.fit(X_train_1, y_train)
y_pred_1 = model_1.predict(X_true_1)

plt.plot(x_true, y_pred_1, 'r--', linewidth=2, label='Linear fit (degree 1)')
plt.title('🔴 HIGH BIAS (Underfitting)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

# Model 2: Just right (degree 3) - BALANCED
plt.subplot(1, 3, 2)
plt.scatter(x_train, y_train, alpha=0.7, color='blue', label='Training data')
plt.plot(x_true, y_true, 'g-', linewidth=2, label='True function')

poly_3 = PolynomialFeatures(degree=3)
X_train_3 = poly_3.fit_transform(x_train.reshape(-1, 1))
X_true_3 = poly_3.transform(x_true.reshape(-1, 1))
model_3 = LinearRegression()
model_3.fit(X_train_3, y_train)
y_pred_3 = model_3.predict(X_true_3)

plt.plot(x_true, y_pred_3, 'orange', linewidth=2, label='Polynomial fit (degree 3)')
plt.title('🟢 BALANCED (Just Right)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

# Model 3: Too complex (degree 15) - HIGH VARIANCE
plt.subplot(1, 3, 3)
plt.scatter(x_train, y_train, alpha=0.7, color='blue', label='Training data')
plt.plot(x_true, y_true, 'g-', linewidth=2, label='True function')

poly_15 = PolynomialFeatures(degree=15)
X_train_15 = poly_15.fit_transform(x_train.reshape(-1, 1))
X_true_15 = poly_15.transform(x_true.reshape(-1, 1))
model_15 = LinearRegression()
model_15.fit(X_train_15, y_train)
y_pred_15 = model_15.predict(X_true_15)

plt.plot(x_true, y_pred_15, 'purple', linewidth=2, label='Polynomial fit (degree 15)')
plt.title('🔴 HIGH VARIANCE (Overfitting)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("📊 WHAT YOU SEE:")
print("   🔴 Left: Too simple → misses the curve (HIGH BIAS)")
print("   🟢 Middle: Just right → captures the pattern (BALANCED)")
print("   🔴 Right: Too complex → fits the noise (HIGH VARIANCE)")
print()
print("🎯 KEY INSIGHT:")
print("   As model complexity increases:")
print("   - Bias decreases (fits training data better)")
print("   - Variance increases (more sensitive to data changes)")
print("   - Sweet spot is in the middle!")
```

```python
## 🛡️ Regularization: Ridge and Lasso Regression

print("🛡️ REGULARIZATION: The Solution to Overfitting")
print()
print("🤔 THE PROBLEM:")
print("   High-degree polynomials overfit (high variance)")
print("   But we still want flexible models!")
print()
print("💡 THE SOLUTION:")
print("   Add a PENALTY for complex models")
print("   Force the model to be simpler!")
print()
print("🎯 TWO MAIN TYPES:")
print("   1. Ridge Regression (L2 penalty)")
print("   2. Lasso Regression (L1 penalty)")

# Generate data with many features (some irrelevant)
np.random.seed(42)
n_samples, n_features = 100, 20
X_reg = np.random.randn(n_samples, n_features)

# Only first 5 features are actually important
true_coef = np.zeros(n_features)
true_coef[:5] = [1.5, -2.0, 0.5, -1.0, 3.0]

y_reg = X_reg @ true_coef + np.random.randn(n_samples) * 0.1

# Split data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.3, random_state=42)

# Compare different models
models = {
    'Linear Regression': LinearRegression(),
    'Ridge (α=1.0)': Ridge(alpha=1.0),
    'Lasso (α=0.1)': Lasso(alpha=0.1)
}

plt.figure(figsize=(15, 5))

for i, (name, model) in enumerate(models.items()):
    plt.subplot(1, 3, i+1)
    
    # Fit model
    model.fit(X_train_reg, y_train_reg)
    
    # Plot coefficients
    if hasattr(model, 'coef_'):
        coef = model.coef_
    else:
        coef = model.coef_
    
    plt.bar(range(len(coef)), coef, alpha=0.7)
    plt.axhline(y=0, color='black', linestyle='-', alpha=0.3)
    
    # Highlight true non-zero coefficients
    for j in range(5):
        plt.bar(j, coef[j], alpha=0.9, color='red')
    
    plt.title(f'{name}')
    plt.xlabel('Feature Index')
    plt.ylabel('Coefficient Value')
    plt.grid(True, alpha=0.3)
    
    # Calculate and display scores
    train_score = model.score(X_train_reg, y_train_reg)
    test_score = model.score(X_test_reg, y_test_reg)
    plt.text(0.02, 0.98, f'Train R²: {train_score:.3f}\nTest R²: {test_score:.3f}', 
             transform=plt.gca().transAxes, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.tight_layout()
plt.show()

print("🔍 WHAT YOU SEE:")
print("   🔴 Red bars = True important features (first 5)")
print("   🔵 Blue bars = Irrelevant features (should be zero)")
print()
print("📊 COMPARISON:")
print("   • Linear Regression: Uses all features (overfits)")
print("   • Ridge: Shrinks all coefficients (reduces overfitting)")
print("   • Lasso: Sets some coefficients to zero (feature selection)")
print()
print("✅ Notice how Lasso automatically finds the important features!")
```

```python
## 🔍 Ridge vs Lasso: When to Use Which?

print("🔍 RIDGE vs LASSO: The Ultimate Comparison")
print()
print("🏔️ RIDGE REGRESSION (L2 Penalty):")
print("   • Penalty: α × Σ(βᵢ²)")
print("   • Effect: Shrinks coefficients toward zero")
print("   • Keeps all features (none become exactly zero)")
print("   • Good when: All features are somewhat useful")
print()
print("🎯 LASSO REGRESSION (L1 Penalty):")
print("   • Penalty: α × Σ|βᵢ|")
print("   • Effect: Sets some coefficients to exactly zero")
print("   • Automatic feature selection")
print("   • Good when: Many features are irrelevant")
print()
print("⚖️ THE PARAMETER α (alpha):")
print("   • α = 0: No regularization (regular linear regression)")
print("   • α small: Little regularization")
print("   • α large: Strong regularization (simpler model)")
print("   • Use cross-validation to find best α!")

# Demonstrate effect of different alpha values
alphas = [0.001, 0.01, 0.1, 1.0, 10.0]

plt.figure(figsize=(15, 10))

# Ridge regression with different alphas
plt.subplot(2, 1, 1)
for alpha in alphas:
    ridge = Ridge(alpha=alpha)
    ridge.fit(X_train_reg, y_train_reg)
    test_score = ridge.score(X_test_reg, y_test_reg)
    plt.plot(range(len(ridge.coef_)), ridge.coef_, 'o-', 
             label=f'α={alpha}, Test R²={test_score:.3f}', alpha=0.7)

plt.title('🏔️ Ridge Regression: Effect of α')
plt.xlabel('Feature Index')
plt.ylabel('Coefficient Value')
plt.legend()
plt.grid(True, alpha=0.3)

# Lasso regression with different alphas
plt.subplot(2, 1, 2)
for alpha in alphas:
    lasso = Lasso(alpha=alpha, max_iter=1000)
    lasso.fit(X_train_reg, y_train_reg)
    test_score = lasso.score(X_test_reg, y_test_reg)
    plt.plot(range(len(lasso.coef_)), lasso.coef_, 'o-', 
             label=f'α={alpha}, Test R²={test_score:.3f}', alpha=0.7)

plt.title('🎯 Lasso Regression: Effect of α')
plt.xlabel('Feature Index')
plt.ylabel('Coefficient Value')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("📊 KEY OBSERVATIONS:")
print("   • As α increases → coefficients get smaller")
print("   • Ridge: Coefficients approach zero but never reach it")
print("   • Lasso: Coefficients become exactly zero (feature selection)")
print("   • Sweet spot: Balance between fit and simplicity")
print()
print("🎯 EXAM TIP:")
print("   Ridge = Shrinkage, Lasso = Selection")
```

# 🔍 LECTURE 8: K-MEANS CLUSTERING

## From Supervised to Unsupervised Learning

print("🔍 LECTURE 8: K-MEANS CLUSTERING")
print("Finding hidden patterns in data!")
print()
print("🧠 BIG SHIFT:")
print("   So far: Supervised learning (we had labels)")
print("   Now: Unsupervised learning (no labels!)")
print()
print("🎯 THE GOAL:")
print("   Find groups (clusters) of similar data points")
print("   WITHOUT being told what the groups are!")
print()
print("🌟 REAL-WORLD EXAMPLES:")
print("   • Customer segmentation (find customer types)")
print("   • Gene analysis (group similar genes)")
print("   • Image compression (group similar colors)")
print("   • Market research (find consumer groups)")


```python
## 🎯 K-Means Algorithm: Step by Step

print("🎯 K-MEANS ALGORITHM: How It Works")
print()
print("📋 THE ALGORITHM (4 Simple Steps):")
print("   1. Choose K (number of clusters)")
print("   2. Place K random centroids")
print("   3. Assign each point to nearest centroid")
print("   4. Move centroids to center of their points")
print("   5. Repeat steps 3-4 until convergence")
print()
print("🧠 KEY INSIGHT:")
print("   It's like finding the 'center' of each group!")

# Generate sample clustering data
np.random.seed(42)
X_cluster, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)

print(f"\n📊 Generated {len(X_cluster)} data points")
print(f"   We know there are 4 true clusters")
print(f"   But let's pretend we don't know that!")

# Visualize the algorithm steps
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.ravel()

# Step 0: Original data (no clustering)
axes[0].scatter(X_cluster[:, 0], X_cluster[:, 1], alpha=0.7, s=50)
axes[0].set_title('🔍 Step 0: Original Data\n(No clustering yet)')
axes[0].grid(True, alpha=0.3)

# Step 1: Random initialization
kmeans_init = KMeans(n_clusters=4, init='random', n_init=1, max_iter=1, random_state=42)
kmeans_init.fit(X_cluster)

axes[1].scatter(X_cluster[:, 0], X_cluster[:, 1], alpha=0.7, s=50)
axes[1].scatter(kmeans_init.cluster_centers_[:, 0], kmeans_init.cluster_centers_[:, 1], 
                c='red', marker='x', s=200, linewidths=3, label='Initial centroids')
axes[1].set_title('🎯 Step 1: Random Centroids\n(Red X marks)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Steps 2-5: Show iterations
for iteration in [1, 2, 5, 10]:
    kmeans_iter = KMeans(n_clusters=4, init='random', n_init=1, max_iter=iteration, random_state=42)
    labels = kmeans_iter.fit_predict(X_cluster)
    
    ax_idx = iteration + 1 if iteration <= 2 else (3 if iteration == 5 else 4)
    
    # Color points by cluster
    colors = ['red', 'blue', 'green', 'purple']
    for i in range(4):
        mask = labels == i
        axes[ax_idx].scatter(X_cluster[mask, 0], X_cluster[mask, 1], 
                           c=colors[i], alpha=0.7, s=50, label=f'Cluster {i}')
    
    # Plot centroids
    axes[ax_idx].scatter(kmeans_iter.cluster_centers_[:, 0], kmeans_iter.cluster_centers_[:, 1], 
                        c='black', marker='x', s=200, linewidths=3, label='Centroids')
    
    axes[ax_idx].set_title(f'🔄 After {iteration} iteration{"s" if iteration > 1 else ""}\n(Colors = clusters)')
    axes[ax_idx].legend()
    axes[ax_idx].grid(True, alpha=0.3)

# Final result
kmeans_final = KMeans(n_clusters=4, random_state=42)
labels_final = kmeans_final.fit_predict(X_cluster)

for i in range(4):
    mask = labels_final == i
    axes[5].scatter(X_cluster[mask, 0], X_cluster[mask, 1], 
                   c=colors[i], alpha=0.7, s=50, label=f'Cluster {i}')

axes[5].scatter(kmeans_final.cluster_centers_[:, 0], kmeans_final.cluster_centers_[:, 1], 
                c='black', marker='x', s=200, linewidths=3, label='Final centroids')
axes[5].set_title('✅ Final Result\n(Converged!)')
axes[5].legend()
axes[5].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("🎯 WHAT YOU SEE:")
print("   • Algorithm starts with random centroids")
print("   • Points get assigned to nearest centroid")
print("   • Centroids move to center of their points")
print("   • Process repeats until centroids stop moving")
print("   • Final result: Clear, separated clusters!")
```

```python
## ❓ The Big Question: How Many Clusters (K)?

print("❓ THE BIG QUESTION: How Many Clusters?")
print()
print("🤔 THE PROBLEM:")
print("   K-means requires you to specify K beforehand")
print("   But how do you know the right number?")
print()
print("💡 THE SOLUTION: Try different K values!")
print("   Use the 'Elbow Method' to find the best K")

# Demonstrate elbow method
K_range = range(1, 11)
inertias = []

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_cluster)
    inertias.append(kmeans.inertia_)  # Sum of squared distances to centroids

plt.figure(figsize=(12, 5))

# Left plot: Elbow method
plt.subplot(1, 2, 1)
plt.plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (Sum of Squared Distances)')
plt.title('📈 Elbow Method')
plt.grid(True, alpha=0.3)

# Highlight the elbow
plt.axvline(x=4, color='red', linestyle='--', alpha=0.7, label='Elbow at K=4')
plt.legend()

# Right plot: Different K values
plt.subplot(1, 2, 2)
k_values_to_show = [2, 3, 4, 5]
colors = ['red', 'blue', 'green', 'purple', 'orange']

for i, k in enumerate(k_values_to_show):
    plt.subplot(2, 2, i+1)
    
    kmeans_k = KMeans(n_clusters=k, random_state=42)
    labels_k = kmeans_k.fit_predict(X_cluster)
    
    for j in range(k):
        mask = labels_k == j
        plt.scatter(X_cluster[mask, 0], X_cluster[mask, 1], 
                   c=colors[j], alpha=0.7, s=30)
    
    plt.scatter(kmeans_k.cluster_centers_[:, 0], kmeans_k.cluster_centers_[:, 1], 
                c='black', marker='x', s=100, linewidths=2)
    plt.title(f'K = {k}')
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("📊 ELBOW METHOD EXPLAINED:")
print("   • Plot inertia (total distance to centroids) vs K")
print("   • Look for the 'elbow' - where improvement slows down")
print("   • K=4 is the elbow point (matches true number of clusters!)")
print()
print("🎯 PRACTICAL TIPS:")
print("   • Try K from 1 to 10 (or more)")
print("   • Look for sharp drops in inertia")
print("   • Consider domain knowledge too")
print("   • Sometimes the elbow isn't perfectly clear")
```

```python
## 🏋️ Practice Problem: Customer Segmentation

print("🏋️ PRACTICE PROBLEM: Customer Segmentation")
print()
print("🎯 SCENARIO:")
print("   You work for an e-commerce company")
print("   You have data on customer spending and frequency")
print("   Goal: Find different customer types for targeted marketing")
print()
print("📋 YOUR TASKS:")
print("   1. Generate synthetic customer data")
print("   2. Use elbow method to find optimal K")
print("   3. Apply K-means clustering")
print("   4. Interpret the customer segments")

# Generate customer data
np.random.seed(42)

# Create 3 customer types
# Type 1: Low spenders, low frequency
customers_1 = np.random.multivariate_normal([20, 2], [[50, 10], [10, 1]], 100)

# Type 2: Medium spenders, medium frequency  
customers_2 = np.random.multivariate_normal([100, 8], [[200, 20], [20, 4]], 150)

# Type 3: High spenders, high frequency
customers_3 = np.random.multivariate_normal([300, 20], [[500, 50], [50, 16]], 100)

# Combine all customers
customer_data = np.vstack([customers_1, customers_2, customers_3])
spending = customer_data[:, 0]
frequency = customer_data[:, 1]

# Ensure positive values
spending = np.maximum(spending, 1)
frequency = np.maximum(frequency, 0.1)

plt.figure(figsize=(15, 5))

# Original data
plt.subplot(1, 3, 1)
plt.scatter(spending, frequency, alpha=0.7, s=50)
plt.xlabel('Monthly Spending ($)')
plt.ylabel('Purchase Frequency (times/month)')
plt.title('🛍️ Customer Data\n(No clustering yet)')
plt.grid(True, alpha=0.3)

# Elbow method
customer_features = np.column_stack([spending, frequency])
K_range = range(1, 8)
inertias = []

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(customer_features)
    inertias.append(kmeans.inertia_)

plt.subplot(1, 3, 2)
plt.plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('📈 Elbow Method\n(Find optimal K)')
plt.axvline(x=3, color='red', linestyle='--', alpha=0.7, label='Elbow at K=3')
plt.legend()
plt.grid(True, alpha=0.3)

# Final clustering with K=3
plt.subplot(1, 3, 3)
kmeans_customers = KMeans(n_clusters=3, random_state=42)
customer_labels = kmeans_customers.fit_predict(customer_features)

colors = ['red', 'blue', 'green']
segment_names = ['Budget Buyers', 'Regular Customers', 'VIP Customers']

for i in range(3):
    mask = customer_labels == i
    plt.scatter(spending[mask], frequency[mask], 
               c=colors[i], alpha=0.7, s=50, label=segment_names[i])

plt.scatter(kmeans_customers.cluster_centers_[:, 0], 
           kmeans_customers.cluster_centers_[:, 1], 
           c='black', marker='x', s=200, linewidths=3, label='Centroids')

plt.xlabel('Monthly Spending ($)')
plt.ylabel('Purchase Frequency (times/month)')
plt.title('🎯 Customer Segments\n(K=3 clusters)')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Analyze the segments
print("📊 CUSTOMER SEGMENT ANALYSIS:")
for i in range(3):
    mask = customer_labels == i
    avg_spending = spending[mask].mean()
    avg_frequency = frequency[mask].mean()
    count = mask.sum()
    
    print(f"\n🏷️ {segment_names[i]}:")
    print(f"   • Count: {count} customers ({count/len(customer_labels)*100:.1f}%)")
    print(f"   • Avg spending: ${avg_spending:.2f}/month")
    print(f"   • Avg frequency: {avg_frequency:.1f} purchases/month")

print(f"\n✅ BUSINESS INSIGHTS:")
print(f"   • Budget Buyers: Price-sensitive, infrequent shoppers")
print(f"   • Regular Customers: Main customer base, moderate spending")
print(f"   • VIP Customers: High-value customers, frequent buyers")
print(f"   • Marketing strategy: Different campaigns for each segment!")
```
